```python
import pandas as pd
import seaborn as sns

t = sns.load_dataset('titanic')
```

```python
# Q1: checking total rows and columns
print("Shape of dataset:", t.shape)
print("Rows:", t.shape[0])
print("Columns:", t.shape[1])
```

```python
# Q2: printing all column names and describing 3 columns
print(t.columns)

# 1. 'survived': stores whether the passenger survived or not (0 for no, 1 for yes)
# 2. 'pclass': ticket class of the passenger (1st, 2nd, or 3rd class)
# 3. 'fare': the ticket price / amount of money paid for the journey
```

```python
# Q3: first 7 rows
t.head(7)
```

```python
# Q3: last 3 rows
t.tail(3)
```

```python
# Q4: checking info and missing values
t.info()

# Total rows = 891
# Columns with missing values:
# - age: 714 non-null -> 177 missing values (891 - 714)
# - embarked: 889 non-null -> 2 missing values (891 - 889)
# - embark_town: 889 non-null -> 2 missing values (891 - 889)
# - deck: 203 non-null -> 688 missing values (891 - 203)
```

```python
# Q5: summary statistics and checking excluded columns
t.describe()

# Two columns left out: 'sex' and 'embarked'
# Reason: describe() by default only includes numerical columns (int/float).
# Text/categorical columns don't have mathematical summary metrics like mean, std, or quartiles.
```

```python
# Q6: analyzing 'class' column using 3 different functions

# a) how many different values
print("Unique count:", t['class'].nunique())

# b) what the values are
print("Unique values:", t['class'].unique())

# c) count of passengers in each class
print(t['class'].value_counts())
```

```python
# Q7: male vs female passenger count
t['sex'].value_counts()
```

```python
# Q8: comparing age mean and median (50%)
t['age'].describe()

# Mean = ~29.70, Median (50%) = 28.00
# The mean is slightly larger than the median.
# This indicates a slight right skew, meaning a few older passengers are pulling the average age upward.
```

```python
# Q9: percentage of missing age values using shape and info
total_rows = t.shape[0]     # 891 from shape
non_null_age = 714          # from info()

missing_age = total_rows - non_null_age
missing_percentage = (missing_age / total_rows) * 100

print(f"Missing Age Percentage: {missing_percentage:.2f}%")

# About 19.87% of age values are missing
```
