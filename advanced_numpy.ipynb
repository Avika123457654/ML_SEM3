```python
# Q1: broadcasting bonus, clipping > 100, and counting capped marks
marks = np.array([[88, 96, 95],
                  [72, 99, 91],
                  [98, 85, 97],
                  [60, 75, 80]])

bonus = np.array([5, 2, 8])

# adding bonus using broadcasting in one line
total_marks = marks + bonus

# counting how many marks exceeded 100
capped_count = np.sum(total_marks > 100)

# capping max marks to 100
final_marks = np.clip(total_marks, a_min=None, a_max=100)

print("Final Marks:\n", final_marks)
print("Total marks capped:", capped_count)

```

```python
# Q4: joining arrays with hstack vs proper 2-column table
heights = np.array([172, 158, 181, 166])
weights = np.array([68, 52, 84, 59])

# first attempt with hstack
first_attempt = np.hstack((heights, weights))
print("First attempt shape:", first_attempt.shape)

# Why it did not produce two columns:
# heights and weights are 1D arrays, so hstack joins them along the only existing axis into one flat array of length 8.

# proper (4, 2) table
table = np.column_stack((heights, weights))
print("Proper table:\n", table)
print("Table shape:", table.shape)
```

```python
# Q8: column-wise min-max normalization
X = np.array([[ 50, 1200, 3],
              [ 80, 1500, 9],
              [ 65,  900, 5],
              [ 95, 2000, 7]])

col_min = X.min(axis=0)
col_max = X.max(axis=0)

X_norm = (X - col_min) / (col_max - col_min)

print("Normalized matrix:\n", X_norm)
print("Column mins:", X_norm.min(axis=0))
print("Column maxs:", X_norm.max(axis=0))
```

```python
# Q9: awarding medals using nested np.where and counting them
scores = np.array([94, 61, 77, 45, 90, 88, 59, 72])

medals = np.where(scores >= 90, 'Gold',
         np.where(scores >= 75, 'Silver',
         np.where(scores >= 60, 'Bronze', 'None')))

medal_names, counts = np.unique(medals, return_counts=True)

print("Assigned medals:\n", medals)
print("\nMedal counts:")
for name, count in zip(medal_names, counts):
    print(f"- {name}: {count}")
```

```python
# Q10: finding elements > 50 and replacing in a copy
m = np.array([[10, 80, 45],
              [55, 20, 99],
              [70, 33, 12]])

# getting row and column indices
rows, cols = np.where(m > 50)

print("Elements greater than 50:", m[rows, cols])
print("Row indices:", rows)
print("Column indices:", cols)

# creating a copy to modify
m_copy = m.copy()
m_copy[rows, cols] = 0

print("\nModified copy:\n", m_copy)
print("Original matrix (unchanged):\n", m)
```

```python
# Q13: matrix multiplication shapes and legality check
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8, 9],
              [1, 0, 2]])

res1 = A @ B.T
res2 = A.T @ B

print("A @ B.T shape:", res1.shape)
print("A.T @ B shape:", res2.shape)

# Explanation:
# - A is (2, 3) and B.T is (3, 2). Inner dimensions match (3 == 3), so output is (2, 2).
# - A.T is (3, 2) and B is (2, 3). Inner dimensions match (2 == 2), so output is (3, 3).
# - A @ B is not allowed because for (2, 3) @ (2, 3), the inner dimensions 3 and 2 do not match.
```

```python
# Q14: multi-step pipeline with intermediate shapes
hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])

# step 1: converting to column vectors
h_col = hours.reshape(-1, 1)
a_col = attendance.reshape(-1, 1)
print("Step 1 (Column vectors shapes):", h_col.shape, a_col.shape)

# step 2: 2-column table
t2 = np.hstack((h_col, a_col))
print("Step 2 (Two-column table shape):", t2.shape)

# step 3: adding row total column
totals = t2.sum(axis=1, keepdims=True)
t3 = np.hstack((t2, totals))
print("Step 3 (Three-column table shape):", t3.shape)

# step 4: multiply with weights
w = np.array([0.5, 0.2, 0.3])
scores = t3 @ w
print("Step 4 (Scores shape):", scores.shape)

# step 5: flag above average
avg = scores.mean()
flags = np.where(scores > avg, 1, 0)
print("Step 5 (Final flags):", flags)
print("Step 5 (Flags shape):", flags.shape)
```

```python
# Q15: min-max normalization function
def min_max_scaler(arr):
    return (arr - np.min(arr)) / (np.max(arr) - np.min(arr))

test_arr = np.array([10, 25, 40, 75, 100])
print("Original:", test_arr)
print("Normalized:", min_max_scaler(test_arr))
```

```python
# Q16: comparing flatten() and ravel()
m = np.array([[4, 8, 15],
              [16, 23, 42]])

flat = m.flatten()
rav = m.ravel()

flat[0] = 999
rav[0] = 888

print("Original matrix:\n", m)

# Explanation:
# - flatten() returns a deep copy, so changing flat[0] has no effect on the original matrix.
# - ravel() returns a view of the original array, so modifying rav[0] directly changed m[0, 0] to 888.
```

```python
# Q17: weekly temperature readings analysis
np.random.seed(7)
temps = np.random.randint(15, 41, 28)

# reshaping into 4 weeks of 7 days
weeks = temps.reshape(4, 7)
print("Weeks shape:", weeks.shape)

week_avgs = weeks.mean(axis=1)
hottest_week = np.argmax(week_avgs)
hot_days = np.sum(temps > 35)

print("Weekly averages:", np.round(week_avgs, 2))
print(f"Hottest week: Week {hottest_week + 1} (index {hottest_week}) with average {week_avgs[hottest_week]:.2f}°C")
print("Total days crossing 35°C:", hot_days)
```

```python
# Q18: matrix multiplication order check
np.random.seed(42)
A = np.random.rand(3, 3)
B = np.random.rand(3, 3)

AB = A @ B
BA = B @ A

print("Is A @ B equal to B @ A?:", np.array_equal(AB, BA))

# Conclusion:
# Matrix multiplication is not commutative (A @ B != B @ A in general).
```

```python
# Q19: replacing failing marks, calculating student totals, and finding topper
marks = np.array([[45, 28, 67, 80],
                  [33, 90, 55, 41],
                  [72, 66, 30, 58],
                  [88, 79, 91, 84],
                  [25, 40, 38, 52]])

# replacing marks below 35 with 35
updated_marks = np.where(marks < 35, 35, marks)

# computing totals per student
student_totals = updated_marks.sum(axis=1)
topper_idx = np.argmax(student_totals)

print("Updated marks:\n", updated_marks)
print("Totals:", student_totals)
print(f"Topper is student at index {topper_idx} with total marks {student_totals[topper_idx]}")
```

```python
# Q20: plain slice (view) vs copy()
arr = np.arange(1, 11)

s_view = arr[2:6]
s_copy = arr[2:6].copy()

s_view[0] = 999
s_copy[0] = 888

print("Original array:", arr)

# Explanation:
# The change made to s_view survived in the original array because a basic slice creates a view (shares the same memory buffer).
# The change to s_copy did not affect the original array because .copy() creates an independent array in memory.
```

```python
# Q21: extracting diagonal vs building diagonal matrix with np.diag
mat_4x4 = np.array([[10,  2,  3,  4],
                    [ 5, 20,  7,  8],
                    [ 9,  1, 30,  3],
                    [ 4,  5,  6, 40]])

# 1. Extracting diagonal from 2D matrix
diag_values = np.diag(mat_4x4)
print("Extracted diagonal:", diag_values)
print("Extracted diagonal shape:", diag_values.shape)

# 2. Building diagonal matrix from 1D array
diag_matrix = np.diag(diag_values)
print("\nBuilt diagonal matrix:\n", diag_matrix)
print("Diagonal matrix shape:", diag_matrix.shape)
```
