```python
# Q2: printing all column names
print(t.columns)

# 1. 'survived' - whether the passenger survived or died (0 = no, 1 = yes)
# 2. 'pclass' - ticket class like 1st, 2nd, or 3rd class
# 3. 'fare' - how much money the passenger paid for their ticket
```

```python
# Q3: first 7 rows
t.head(7) 
# Q3: last 3 rows
t.tail(3)
```

```python
# Q4: checking column info and missing values
t.info()

# total rows = 891
# columns with missing values:
# - age: 714 non-null -> 177 values missing (891 - 714)
# - embarked: 889 non-null -> 2 values missing (891 - 889)
# - embark_town: 889 non-null -> 2 values missing (891 - 889)
# - deck: 203 non-null -> 688 values missing (891 - 203)
```

```python
# Q5: summary stats
t.describe()

# Two columns left out are 'sex' and 'embarked'.
# This is because describe() by default only works on numeric data (int/float).
# Categorical/text columns don't have mathematical values like mean or standard deviation.
```

```python
# Q6: class column using 3 different functions

# a) number of unique values
print("Unique count:", t['class'].nunique())

# b) what the values are
print("Classes:", t['class'].unique())

# c) count of passengers in each class
print(t['class'].value_counts())
```

```python
# Q7: male vs female count
t['sex'].value_counts()

# male: 577, female: 314
```

```python
# Q8: comparing age mean and median (50%)
t['age'].describe()

# Mean = 29.70, 50% (Median) = 28.00
# The mean is slightly higher than the median.
# This means the age distribution is slightly right skewed because older passengers are pulling the average up.
```

```python
# Q9: missing age percentage using values from shape and info
total_rows = 891        # from t.shape
non_null_age = 714      # from t.info()

missing_age = total_rows - non_null_age
missing_pct = (missing_age / total_rows) * 100

print(f"Missing Age Percentage: {missing_pct:.2f}%")

# About 19.87% of the age values are missing.
```
