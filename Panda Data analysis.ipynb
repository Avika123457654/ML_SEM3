```python
import pandas as pd
import seaborn as sns

# loading datasets
tips = sns.load_dataset('tips')
# or: tips = pd.read_csv('tips.csv')

titanic = sns.load_dataset('titanic')
# or: titanic = pd.read_csv('Titanic-Dataset.csv')
```

```python
# A1: calculating tip_percent rounded to 1 decimal place and showing first 5 rows
tips['tip_percent'] = round((tips['tip'] / tips['total_bill']) * 100, 1)
tips[['total_bill', 'tip', 'tip_percent']].head()
```

```python
# A2: using positions only (iloc) - no column names
# 1. total_bill of 10th row (index 9, column 0)
print("10th row total_bill:", tips.iloc[9, 0])

# 2. last 3 rows with only first two columns
tips.iloc[-3:, :2]
```

```python
# A3: using labels only (loc) for rows 20 to 25 and columns sex through size
tips.loc[20:25, 'sex':'size']

# Question: How many rows came back, and why is it not 5?
# Answer: 6 rows came back (rows 20, 21, 22, 23, 24, 25).
# This is because loc is label-based and includes both the start and end labels.
```

```python
# A4: boolean indexing counts

# 1. bills not at Dinner using ~
not_dinner_count = len(tips[~(tips['time'] == 'Dinner')])
print("Bills not at Dinner:", not_dinner_count)

# 2. party >= 3 and tip > 4
large_party_good_tip = len(tips[(tips['size'] >= 3) & (tips['tip'] > 4)])
print("Party >= 3 and tip > 4:", large_party_good_tip)
```

```python
# A5: flagging generous customers and counting
tips['service'] = 'normal'
tips.loc[tips['tip_percent'] > 20, 'service'] = 'generous'

tips['service'].value_counts()
```

```python
# B1: passengers boarded from each port
# If using Titanic-Dataset.csv with original capitalization:
# (change 'Embarked' to 'embarked' if using seaborn dataset)
col_embarked = 'Embarked' if 'Embarked' in titanic.columns else 'embarked'
print(titanic[col_embarked].value_counts())

# Adding the three numbers up:
# S (644) + C (168) + Q (77) = 889
# They do not come to 891 because 2 passengers have missing (NaN) values in the Embarked column.
```

```python
# B2: female passenger counts in third class
col_sex = 'Sex' if 'Sex' in titanic.columns else 'sex'
col_pclass = 'Pclass' if 'Pclass' in titanic.columns else 'pclass'
col_survived = 'Survived' if 'Survived' in titanic.columns else 'survived'

# 1. women in third class
women_3rd = titanic[(titanic[col_sex] == 'female') & (titanic[col_pclass] == 3)]
print("Women in 3rd class:", len(women_3rd))

# 2. women in third class who survived
women_3rd_survived = titanic[(titanic[col_sex] == 'female') & (titanic[col_pclass] == 3) & (titanic[col_survived] == 1)]
print("Women in 3rd class who survived:", len(women_3rd_survived))
```

```python
# B3: passengers who paid fare > 200
col_name = 'Name' if 'Name' in titanic.columns else 'who'
col_age = 'Age' if 'Age' in titanic.columns else 'age'
col_fare = 'Fare' if 'Fare' in titanic.columns else 'fare'

expensive_tickets = titanic.loc[titanic[col_fare] > 200, [col_name, col_age, col_fare]]
print(expensive_tickets)
print("Total passengers paying > 200:", len(expensive_tickets))
```

```python
# B4: survival percentage in first class vs third class
first_class = titanic[titanic[col_pclass] == 1]
third_class = titanic[titanic[col_pclass] == 3]

p1_survival = (first_class[col_survived].sum() / len(first_class)) * 100
p3_survival = (third_class[col_survived].sum() / len(third_class)) * 100

print(f"1st Class Survival: {p1_survival:.2f}%")
print(f"3rd Class Survival: {p3_survival:.2f}%")

# First Class Survival: 62.96% | Third Class Survival: 24.24%
```

```python
# B5: age group classification and handling missing ages
titanic['age_group'] = 'adult'
titanic.loc[titanic[col_age] < 18, 'age_group'] = 'child'

print(titanic['age_group'].value_counts())

# Think before tomorrow:
# The 177 passengers with missing ages landed in 'adult' because NaN < 18 evaluates to False.
# This is not accurate since some of them could be children.
# I would mark missing ages as 'Unknown' or fill/impute the missing ages using median age before assigning groups.
```
