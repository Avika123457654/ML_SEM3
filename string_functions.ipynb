```python
# Q1: cleaning city name and validating characters
city_input = "   nEw  delhi  "

# strip extra spaces and convert to title case
cleaned_city = " ".join(city_input.strip().split()).title()

# validation: check if characters are alphabets or spaces only
is_valid = True
for ch in cleaned_city:
    if not (ch.isalpha() or ch.isspace()):
        is_valid = False
        break

if is_valid and len(cleaned_city) > 0:
    print(cleaned_city)
else:
    print("Invalid city")
```

```python
# Q2: processing prices string and calculating stats
price_str = " 120, 45, 300, 80, 95 "

# split by comma, strip spaces, and convert to integers
prices = [int(p.strip()) for p in price_str.split(",") if p.strip()]

total_bill = sum(prices)
highest_price = max(prices)
avg_price = total_bill / len(prices)

print(f"Total  : {total_bill}")
print(f"Highest: {highest_price}")
print(f"Average: {avg_price:.2f}")
```

```python
# Q3: validating username based on rules
raw_username = " Rahul_123 "

# clean spaces and convert to lowercase
u = raw_username.strip().lower()

# checking rules: alphanumeric, starts with alphabet, length between 5 and 12
if u.isalnum() and u[0].isalpha() and 5 <= len(u) <= 12:
    print("Valid username")
else:
    print("Invalid username")
```

```python
# Q4: sentence analysis (word count, words ending with vowel, longest word)
sentence = "Python is a great programming language to learn today"

words = sentence.split()
vowels = ('a', 'e', 'i', 'o', 'u')

# (a) total words
total_words = len(words)
print("Total number of words:", total_words)

# (b) words ending in vowel
ending_vowel_count = 0
for w in words:
    clean_w = w.lower().strip(".,!?;:")
    if clean_w and clean_w.endswith(vowels):
        ending_vowel_count += 1
print("Words ending with vowel:", ending_vowel_count)

# (c) longest word
longest = ""
for w in words:
    clean_w = w.strip(".,!?;:")
    if len(clean_w) > len(longest):
        longest = clean_w
print("Longest word:", longest)
```

```python
# Q5: character frequency ignoring spaces and case
text = "Aba c"
char_counts = {}

for ch in text.lower():
    if ch != " ":
        char_counts[ch] = char_counts.get(ch, 0) + 1

print(char_counts)
```

```python
# Q6: word frequency and finding the most frequent word
sentence = "apple banana apple orange banana apple grapes"
words = sentence.lower().split()

word_counts = {}
for w in words:
    word_counts[w] = word_counts.get(w, 0) + 1

# find most frequent word using sorted with key=lambda
sorted_words = sorted(word_counts.items(), key=lambda x: x[1], reverse=True)
most_frequent_word, highest_freq = sorted_words[0]

print("Word Frequencies:", word_counts)
print(f"Most frequent word: '{most_frequent_word}' (occurred {highest_freq} times)")
```

```python
# Q7: invert dictionary with grouping
sales = {"mon": 200, "tue": 350, "wed": 200}
inverted_sales = {}

for day, val in sales.items():
    if val not in inverted_sales:
        inverted_sales[val] = [day]
    else:
        inverted_sales[val].append(day)

print(inverted_sales)
```

```python
# Q8: merge two test score dictionaries and sum common values
test1 = {"amit": 40, "bina": 35}
test2 = {"amit": 30, "chirag": 45}

combined_marks = test1.copy()

for student, marks in test2.items():
    combined_marks[student] = combined_marks.get(student, 0) + marks

print(combined_marks)
```

```python
# Q9: sort employees by salary descending using key lambda
employees = [("Ravi", 55000), ("Sita", 72000), ("Amit", 48000)]

# sorting by second element of tuple
sorted_employees = sorted(employees, key=lambda emp: emp[1], reverse=True)

for name, salary in sorted_employees:
    print(f"{name} earns Rs. {salary}")
```

```python
# Q10: movie dataset analysis
movies = [
    {"title": "Movie A", "rating": 8.2, "genre": "Thriller"},
    {"title": "Movie B", "rating": 8.4, "genre": "Sports"},
    {"title": "Movie C", "rating": 8.1, "genre": "Comedy"},
    {"title": "Movie D", "rating": 8.2, "genre": "Drama"}
]

# (a) movies with rating > 8.1
print("Movies with rating > 8.1:")
for m in movies:
    if m["rating"] > 8.1:
        print(f"- {m['title']}")

# (b) extract all genres (no duplicates)
unique_genres = list(set(m["genre"] for m in movies))
print("\nGenres:", unique_genres)

# (c) grouping movies by rating
movies_by_rating = {}
for m in movies:
    r = m["rating"]
    t = m["title"]
    if r not in movies_by_rating:
        movies_by_rating[r] = [t]
    else:
        movies_by_rating[r].append(t)

print("\nMovies Grouped by Rating:")
for rating, titles in sorted(movies_by_rating.items(), reverse=True):
    print(f"Rating {rating}: {', '.join(titles)}")
```
