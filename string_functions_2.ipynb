```python
# Question 1: The Strict Username Validator
def validate_username(username):
    cleaned = username.strip()
    if len(cleaned) >= 5 and cleaned.isalnum():
        return cleaned.lower()
    return "Invalid"

print(validate_username("  CodeMaster99  "))
print(validate_username("  Python_Dev  "))
```

```python
# Question 2: The Spy Code Reverser
secret_code = "charlie-bravo-alpha"

words = secret_code.split("-")
words.reverse()
decoded = " ".join(words).upper()

print(decoded)
```

```python
# Question 3: The Messy Database Cleaner
def clean_database(raw_names):
    names = raw_names.split(";")
    cleaned_names = []
    for name in names:
        parts = name.strip().split()
        if parts:
            title_cased = " ".join(parts).title()
            cleaned_names.append(title_cased)
    return ", ".join(cleaned_names)

raw_names = "  john doe  ; jAnE SMiTh ; ALAN tUrInG "
print(clean_database(raw_names))
```

```python
# Question 4: The Multiple Space Eliminator
text = "   python   is   a   very   powerful   language   "

cleaned_words = text.strip().split()
cleaned_sentence = " ".join(cleaned_words).capitalize()

print(cleaned_sentence)
```

```python
# Question 5: The Palindrome Sentence Checker
def is_palindrome_sentence(sentence):
    cleaned = ""
    for ch in sentence.lower():
        if ch.isalnum():
            cleaned += ch
    return cleaned == cleaned[::-1]

print(is_palindrome_sentence("Was it a car or a cat I saw?"))
print(is_palindrome_sentence("Python is fun"))
```

```python
# Question 6: The Hashtag Extractor
post = "Loving the SUNSET! #Nature #Sunset_Vibes #beautiful"

hashtags = []
for word in post.split():
    if word.startswith("#"):
        tag = word.strip("#.,!?:;").lower()
        if tag:
            hashtags.append(tag)

print(hashtags)
```

```python
# Question 7: The File Extension Finder
filenames = "report.pdf, image.JPG, data.csv, notes.txt, photo.jpg"

extensions = []
for file in filenames.split(","):
    cleaned_file = file.strip()
    if "." in cleaned_file:
        ext = cleaned_file.split(".")[-1].upper()
        if ext not in extensions:
            extensions.append(ext)

print(extensions)
```

```python
# Question 8: The Domain Name Slicer
def get_domain_name(email):
    email = email.strip()
    if "@" in email and "." in email:
        domain_part = email.split("@")[1]
        company = domain_part.split(".")[0]
        if company:
            return company.capitalize()
    return "Invalid"

print(get_domain_name("developer_99@microsoft.com"))
print(get_domain_name("admin_at_website.org"))
```

```python
# Question 9: The Anagram Detective
def check_anagram(str1, str2):
    s1 = "".join(ch for ch in str1.lower() if ch.isalpha())
    s2 = "".join(ch for ch in str2.lower() if ch.isalpha())
    return sorted(s1) == sorted(s2)

print(check_anagram("Dormitory", "Dirty room"))
```

```python
# Question 10: The Messy Phone Number Formatter
def format_phone_number(raw_num):
    digits = ""
    for ch in raw_num:
        if ch.isdigit():
            digits += ch
            
    if len(digits) == 10:
        return f"({digits[:3]}) {digits[3:6]}-{digits[6:]}"
    return "Invalid Number"

print(format_phone_number(" +1 (555) 123.4567 "))
print(format_phone_number("555-987-6543"))
```

```python
# Question 11: The Product SKU Inspector
def inspect_sku(raw_sku):
    cleaned = raw_sku.strip().upper()
    
    parts = cleaned.split("-")
    if len(parts) == 2:
        prefix, suffix = parts
        if len(prefix) == 3 and prefix.isalpha() and len(suffix) == 4 and suffix.isdigit():
            return f"Valid: {cleaned}"
            
    return "Invalid SKU"

print(inspect_sku("  xYz-9876 "))
print(inspect_sku("AB-12345"))
print(inspect_sku("A@C-1234"))
```
